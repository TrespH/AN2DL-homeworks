# Import Libraries

```python
import shutil
import os
import zipfile
import sys
import numpy as np
from sklearn.metrics import accuracy_score
```

# 1 - Clear the Extracted Files Directory

```python
extracted_files_dir = 'extracted_files'
if os.path.exists(extracted_files_dir):
    shutil.rmtree(extracted_files_dir)
os.makedirs(extracted_files_dir)
```

# 2 - Extract the zip file 

```python
zip_file_path = 'CNN.zip'
with zipfile.ZipFile(zip_file_path, 'r') as zip_ref:
    zip_ref.extractall(extracted_files_dir)
```

# 3 - Load the model

```python
sys.path.append(extracted_files_dir)
from model import Model
model_instance = Model()
```

# 4 - Load the dataset

```python
data = np.load('training_set.npz')
X = data['images'][:100]
X = X / 255.0
y = data['labels'][:100]
```

```python
y = y.flatten()
```

# 5 - Make predictions

```python
predictions = model_instance.predict(X)
print(f"Predictions: {predictions}")
print(f"Actual labels: {y}")
```

# 6 - Evaluate the predictions

```python
accuracy = accuracy_score(y, predictions)
print(f"Accuracy: {accuracy}")
```
